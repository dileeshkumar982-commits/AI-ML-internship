```python
wine_df = pd.DataFrame(data = wine_data['data'],columns = wine_data['feature_names'])
wine_df
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

wine_df['proline'] = le.fit_transform(wine_df['proline'])

unique_proline = wine_df['proline'].unique()

print("Encoder Species Values:")
print(unique_proline)

```

```python
wine_df
```

```python
fig,axes = plt.subplots(2,2,figsize=(15,10))

sns.histplot(wine_df["alcohol"], kde = True, ax = axes[0,0], color = "purple")
axes[0,0].set_title("Alcohol Distribution")

sns.histplot(wine_df["malic_acid"],kde = True, ax = axes[0,1], color="skyblue")
axes[0,1].set_title("Malic Acid Distribution")

sns.histplot(wine_df["ash"], kde = True, ax=axes[1,0], color="orange")
axes[1,0].set_title("Ash Distribution")

sns.histplot(wine_df["magnesium"], kde = True, ax = axes[1,1], color="red")
axes[1,1].set_title("Magnesium Distribution")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.boxplot(data=wine_df)

plt.title('Boxplot of All Numerical Features (Wine Dataset)')
plt.ylabel('Values')
plt.xticks(rotation=45,ha='right')
plt.tight_layout()

plt.show()
```

```python
from pandas.plotting import andrews_curves
plt.figure(figsize=(10, 6))
andrews_curves(wine_df, 'proline', colormap='viridis')
plt.title('Andrews Curves for wine Dataset')
plt.show()
```

```python
#KEY INSIGHTS

#Dataset characteristics:
#> 178 samples with 13 features each .
#> slightly embalanced classes distributed.
#> no missing values present in the dataset.

#Feature distributions
#> features like ash and alcalinity of ash exhibit the most normal(gaussian) distributions.

#Species differentiation
#> class 0 and class 2 are clearly distinct and easily separable using features like proline and color intensity.

#Correlations
#>flavanoids and total phenols are very highly correlated(around0.86)

#Classification potential
#> flavanoids,proline,and color intensity provide much better classification poer than features like nonflavanoids phenols.

#Conclusion

#This EDA revels that the Wine dataset is highly structured with prominent chemical paters that can be effectively utilized for multiclass classification tasks.Distinct markers like proline levels and flavanoid concentrations are exeptionally usefull for differentiating between specific wine culture.Principal Component Analysis (PCA) or Linear Discrimination Analysis (LDA) before modeling will be highly beneficial.
```
