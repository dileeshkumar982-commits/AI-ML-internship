```python
#HOW TO GET UNIQUE ITEMS AND COUNTS

import numpy as np

arr = np.array([1,2,3,1,2,4,5,3,2,1])
print("original array:",arr)

unique_values = np.unique(arr)
print("unique values:",unique_values)
```

```python
unique_values,counts = np.unique(arr,return_counts = True)
print("\nvalue counts:")
for val, cnt  in  zip (unique_values,counts):
    print(f"{val}:{cnt}")
```

```python
#WORKING WITH MATHEMATICAL FORMULAS

def mse(y_true, y_pred):
    return np.mean((y_true - y_pred)**2)

y_true = np.array([1.0,2.0,3.0,4.0])
y_pred = np.array([0.9,2.1,2.9,4.2])

print("MSE:", mse(y_true,y_pred))

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

x = np.linspace(-5,5,10)
print("\nsigmoid values:",sigmoid(x))

def euclidean(a,b):
    return np.sqrt(np.sum((a - b) ** 2))

point1 = np.array([1,2,3])
point2 = np.array([4,5,6])
print("\neuclidean distance:",euclidean(point1,point2))

x = np.array([[1,1],[1,2],[1,3]])
y = np.array([[1],[2],[3]])
theta = np.linalg.inv(x.T @ x) @ x.T @ y
print("\nlinear regression coefficients:\n",theta)
```
